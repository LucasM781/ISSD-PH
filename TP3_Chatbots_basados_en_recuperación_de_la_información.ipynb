## 2 Crear el chatbot utilizando TF-IDF

```python
# Vectorización con TF-IDF
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(preguntas)

tfidf_matrix
```

```python
# Función para encontrar la mejor coincidencia
def responder_pregunta(consulta_usuario):
    consulta_vec = vectorizer.transform([consulta_usuario])
    similitudes = cosine_similarity(consulta_vec, tfidf_matrix).flatten()
    print(similitudes)
    indice_mejor_coincidencia = similitudes.argmax() # el argumento cuyo valor es el maximo
    print(indice_mejor_coincidencia)
    return respuestas[indice_mejor_coincidencia]
```

```python
df['Vector'] = df['Pregunta'].apply(lambda x: nlp(x).vector)
matriz_vectores_bd = np.vstack(df['Vector'].values)
```

```python
casos_prueba = [
    {
        "input": "El táctil me anda joya pero el cristal de arriba se me astilló todo, tiene arreglo?",
        "indice_esperado": 1, # Pregunta: ¿Pueden cambiar solo el vidrio...?
        "descripcion": "Consulta sobre cambio de visor (Sinónimos exactos)"
    },
    {
        "input": "Se me cayó el celu al inodoro y lo sequé con arroz, q hago?",
        "indice_esperado": 15, # Pregunta: Se me cayó el celular al agua...
        "descripcion": "Consulta sobre celulares mojados (Contexto inferido)"
    },
    {
        "input": "Lo reinicié y se quedó congelado en la manzanita de Apple, no avanza de ahí",
        "indice_esperado": 17, # Pregunta: El celular se queda tildado en el logo...
        "descripcion": "Consulta sobre software / Bootloop"
    },
    {
        "input": "¿Cuánto tiempo demora un cambio de micrófono?",
        "indice_esperado": 11, # Pregunta de audios/micrófono
        "descripcion": "El problema de la 'Estructura Idéntica' (Fallará intencionalmente)"
    },
    {
        "input": "Se me mojó el celular, si le hacen el baño químico pierdo las fotos que tengo?",
        "indice_esperado": 18, # Pregunta sobre borrar datos
        "descripcion": "Conflicto por cruce de temas (Fallará parcialmente)"
    },
    {
        "input": "Hola, ¿tienen fundas para el Motorola G20?",
        "indice_esperado": None, # Falso positivo
        "descripcion": "Consultas fuera de alcance"
    }
]
```

```python
UMBRAL_SIMILITUD = 0.65

print("-" * 50)
print("INICIANDO SIMULACIÓN DEL CHATBOT")
print("-" * 50)

for caso in casos_prueba:
    mensaje = caso["input"]
    indice_esperado = caso["indice_esperado"]

    # Vectorizar input y calcular similitud
    vector_cliente = nlp(mensaje).vector.reshape(1, -1)
    similitudes = cosine_similarity(vector_cliente, matriz_vectores_bd)[0]

    indice_ganador = np.argmax(similitudes)
    puntaje_maximo = similitudes[indice_ganador]

    print(f"\n[{caso['descripcion']}]")
    print(f"Usuario: '{mensaje}'")

    if puntaje_maximo >= UMBRAL_SIMILITUD:
        pregunta_detectada = df.iloc[indice_ganador]['Pregunta']
        respuesta_bot = df.iloc[indice_ganador]['Respuesta']

        print(f"Bot detectó [Índice {indice_ganador}]: '{pregunta_detectada}' (Similitud: {puntaje_maximo:.2f})")

        # Evaluación
        if indice_ganador == indice_esperado:
            print("Resultado: CORRECTO")
        elif indice_esperado is None:
            print("Resultado: FALSO POSITIVO (Respondió cuando no debía)")
        else:
            pregunta_real = df.iloc[indice_esperado]['Pregunta']
            print(f"Resultado: INCORRECTO (Se esperaba: '{pregunta_real}')")
    else:
        print(f"Bot detectó: Nada superó el umbral. (Máxima similitud: {puntaje_maximo:.2f})")
        if indice_esperado is None:
             print("Resultado: CORRECTO (Ignoró la pregunta fuera de alcance)")
        else:
             pregunta_real = df.iloc[indice_esperado]['Pregunta']
             print(f"Resultado: FALSO NEGATIVO (Se esperaba: '{pregunta_real}')")
```

### 4.2 Chatbot vectorizado con embeddings

```python
print("-" * 50)
print("CHATBOT INICIADO. Escribe 'salir' para terminar el programa.")
print("-" * 50)

# 4. Bucle
while True:
    # Recibir input
    mensaje = input("\nTú: ")

    # Condición para romper el bucle
    if mensaje.lower() in ['salir', 'quit', 'exit', 'chau']:
        print("Bot: ¡Nos vemos!")
        break

    # Evitar errores si el usuario presiona Enter sin escribir nada
    if not mensaje.strip():
        continue

    # Vectorizar el mensaje en tiempo real
    vector_cliente = nlp(mensaje).vector.reshape(1, -1)

    # Calcular similitud
    similitudes = cosine_similarity(vector_cliente, matriz_vectores_bd)[0]

    indice_ganador = np.argmax(similitudes)
    puntaje_maximo = similitudes[indice_ganador]

    # Evaluar contra el umbral
    if puntaje_maximo >= UMBRAL_SIMILITUD:
        respuesta_bot = df.iloc[indice_ganador]['Respuesta']
        pregunta_detectada = df.iloc[indice_ganador]['Pregunta']

        # Imprime la respuesta y te muestra (en gris/claro) qué detectó para que depures
        print(f"Bot: {respuesta_bot}")
        print(f"  [Debug - Entendió: '{pregunta_detectada}' con {puntaje_maximo:.2f}]")
    else:
        print("Bot: Disculpa, no entendí tu consulta o no realizamos ese servicio.")
        print(f"  [Debug - Similitud máxima: {puntaje_maximo:.2f}]")
```

## 5 Conclusiones de todo lo realizado

En cuáles respuestas falla o tiene problemas:
- El modelo presenta problemas de "falsos positivos", donde puede intentar responder a consultas que están fuera de alcance (por ejemplo, consultar por fundas de celulares) si el umbral de similitud no está ajustado con precisión.
- Falla intencionalmente ante problemas de "estructura idéntica", donde la oración ingresada es casi igual a una pregunta de la base de datos, pero cambia la entidad o pieza clave.
- Presenta fallos parciales ante un "conflicto por cruce de temas", donde la oración del usuario involucra múltiples intenciones cruzadas.

Cuáles preguntas confunde:

- Confunde consultas que comparten la misma estructura sintáctica general. Por ejemplo, al ingresar "¿Cuánto tiempo demora un cambio de micrófono?", el sistema se confunde con la pregunta almacenada "¿Cuánto tiempo demora un cambio de batería?" porque la mayor parte de los vectores promediados son similares.
- También confunde intenciones en preguntas compuestas, como "Se me mojó el celular, si le hacen el baño químico pierdo las fotos que tengo?". Aquí el modelo se confunde entre la base de conocimiento de daños por agua y la base de retención de datos.

Comparacion en los resultados de los chatbots:

- Chatbot con TF-IDF: Es un modelo basado en la superposición léxica y palabras clave. Funciona excelente cuando el usuario utiliza exactamente el mismo vocabulario de la base de datos. Sin embargo, presenta grandes limitaciones semánticas; si se utilizan expresiones coloquiales, faltas de ortografía o sinónimos (como "celu", "inodoro" o "astilló"), la matriz dispersa no logra encontrar coincidencias y falla.
- Chatbot con Embeddings (spaCy es_core_news_sm): Supera las limitaciones del TF-IDF al lograr capturar el contexto inferido y la similitud semántica de los sinónimos. No obstante, al utilizar un modelo pequeño (sm) y promediar los vectores de toda la oración, diluye el peso de entidades de información críticas. Por ello, aunque entiende mejor el lenguaje natural fluido, es más propenso a confundir temas si la oración es larga o comparte la misma estructura gramatical con otra pregunta distinta.
